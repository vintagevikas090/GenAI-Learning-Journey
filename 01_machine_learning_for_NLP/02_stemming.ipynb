### Stemming and Its Types

```python
from nltk.stem import PorterStemmer, RegexpStemmer, SnowballStemmer
```

#### PorterStemmer

```python
ps = PorterStemmer()
print(ps.stem('finally'), ps.stem('eating'))
```

```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

```python
for word in words:
    print(word, "----->", ps.stem(word, to_lowercase= True))
```

```python
print(ps.stem('fairly'), ps.stem('Congratualations'))
## For certain words -> meaning is getting ruined
```

#### RegexpStemmer

```python
rs = RegexpStemmer(regexp='ing|s$') 
## remove 'ing' from word if present anywhere
## remove 's' from word if word ends with 's'

print('--- RegExpStemmer Output ---')
for word in words:
    print(word, '--->', rs.stem(word))
```

#### SnowballStemmer

```python
sbs = SnowballStemmer('english') # -> language is required Field here
```

```python
print("SnowballStemmer Output")
for word in words:
    print(word, '----> ', sbs.stem(word))
```

```python
sbs.stem('fairly')
```
