#### Bag Of Words

###### Bag of Words (BOW) is a text feature extraction technique in NLP.
###### It converts text into numerical vectors.
###### It creates a vocabulary of unique words from the dataset.
###### Each sentence/document is represented using word frequency counts.
###### BOW does not consider grammar, context, or word order.
###### Commonly used in text classification and sentiment analysis.

```python
corpus = """
Artificial Intelligence and Natural Language Processing are rapidly transforming the 
modern technology landscape. Machine learning models are now capable of understanding 
human language, generating meaningful responses, translating text between languages, and even 
writing computer programs. One of the fundamental tasks in NLP is converting textual data into 
numerical form so that machine learning algorithms can process it efficiently. Traditional techniques 
such as One Hot Encoding, Bag of Words, and TF-IDF are widely used for feature extraction in text analysis. 
The Bag of Words model represents text by counting the occurrence of words in a document without
considering grammar or word order. although simple, BOW is extremely useful for understanding 
important words appear in a dataset. Modern NLP systems go beyond these traditional 
by using embeddings and transformer-based architectures that capture semantic meaning
and contextual relationships between words. Technologies such as neural networks, attention 
mechanisms, and large language models have significantly improved the capabilities of AI systems
in applications like chatbots, recommendation systems, sentiment analysis, document summarization, 
and information retrieval. Ās the field of Generative AI continues to evolve, understanding foundational 
concepts like tokenization, stop word removal, stemming, lemmatization, and feature extraction 
becomes essential for anyone learning NLP and GenAI engineering.

"""
```

##### text preprocessing 

```python
### Required before any feature extraction / vectorization method
```

```python
## removing specail chars 
import re 
corpus = re.sub('[^a-zA-Z]', ' ', corpus)
corpus
```

```python
## lowering the case
corpus = corpus.lower()
```

```python
## apply stopwords
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
sw = stopwords.words('english')

words = word_tokenize(corpus)

words = [word for word in words if word not in sw]

corpus = ' '.join(words)

print(corpus)
```

##### Implement BOW

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer()
```

```python
x = cv.fit_transform(corpus.split()).toarray()
```

```python
print(x)
```

```python
cv.vocabulary_
```

```python
cv1 = CountVectorizer(max_features=50)
```

```python
cv1.fit_transform(corpus.split()).toarray()
```

```python
cv1.vocabulary_
```

##### N GRAMS

###### Helps in capturing the sementics information between the words 

```python
## to understand this, we should Have list of sentences, not words
```

```python
corpus = """
Artificial Intelligence and Natural Language Processing are rapidly transforming the 
modern technology landscape. Machine learning models are now capable of understanding 
human language, generating meaningful responses, translating text between languages, and even 
writing computer programs. One of the fundamental tasks in NLP is converting textual data into 
numerical form so that machine learning algorithms can process it efficiently. Traditional techniques 
such as One Hot Encoding, Bag of Words, and TF-IDF are widely used for feature extraction in text analysis. 
The Bag of Words model represents text by counting the occurrence of words in a document without
considering grammar or word order. although simple, BOW is extremely useful for understanding 
important words appear in a dataset. Modern NLP systems go beyond these traditional 
by using embeddings and transformer-based architectures that capture semantic meaning
and contextual relationships between words. Technologies such as neural networks, attention 
mechanisms, and large language models have significantly improved the capabilities of AI systems
in applications like chatbots, recommendation systems, sentiment analysis, document summarization, 
and information retrieval. Ās the field of Generative AI continues to evolve, understanding foundational 
concepts like tokenization, stop word removal, stemming, lemmatization, and feature extraction 
becomes essential for anyone learning NLP and GenAI engineering.

"""
```

```python
corpus = corpus.lower()
corpus = corpus.split('.')
print(corpus)
```

```python
for i in range(len(corpus)):
    sent = corpus[i].strip()
    sent = [word for word in sent.split() if word not in sw]
    sent = [re.sub('[^a-zA-Z]', " ", word) for word in sent] 
    sent = ' '.join(sent)
    corpus[i] = sent
```

```python
corpus
```

```python
cv2 = CountVectorizer(max_features=70, ngram_range=(2,2))
```

```python
x2 = cv2.fit_transform(corpus)
cv2.vocabulary_
```

```python
cv3 = CountVectorizer(max_features=50, ngram_range=(3,3))
```

```python
cv3.fit_transform(corpus)
cv3.vocabulary_
```
