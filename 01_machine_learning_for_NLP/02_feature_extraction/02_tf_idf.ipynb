### TF-IDF

###### It measures how important a word is in a document relative to the entire corpus.
###### TF (Term Frequency) measures how frequently a word appears in a document.
###### IDF (Inverse Document Frequency) reduces the importance of common words appearing in many documents.
###### Words with high TF-IDF scores are more meaningful for a specific document.
###### Common words like “the”, “is”, and “and” usually get lower scores.

```python
corpus = """
Artificial Intelligence and Natural Language Processing are rapidly transforming the 
modern technology landscape. Machine learning models are now capable of understanding 
human language, generating meaningful responses, translating text between languages, and even 
writing computer programs. One of the fundamental tasks in NLP is converting textual data into 
numerical form so that machine learning algorithms can process it efficiently. Traditional techniques 
such as One Hot Encoding, Bag of Words, and TF-IDF are widely used for feature extraction in text analysis. 
The Bag of Words model represents text by counting the occurrence of words in a document without
considering grammar or word order. although simple, BOW is extremely useful for understanding 
important words appear in a dataset. Modern NLP systems go beyond these traditional 
by using embeddings and transformer-based architectures that capture semantic meaning
and contextual relationships between words. Technologies such as neural networks, attention 
mechanisms, and large language models have significantly improved the capabilities of AI systems
in applications like chatbots, recommendation systems, sentiment analysis, document summarization, 
and information retrieval. Ās the field of Generative AI continues to evolve, understanding foundational 
concepts like tokenization, stop word removal, stemming, lemmatization, and feature extraction 
becomes essential for anyone learning NLP and GenAI engineering.

"""
```

```python
corpus = corpus.split('.')
corpus = [sent.replace('\n', '') for sent in corpus]
corpus = [sent.strip().lower() for sent in corpus]
print(corpus)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tv = TfidfVectorizer()
x = tv.fit_transform(corpus).toarray()
x
```

```python
for i, arr in enumerate(x):
    print(f'ROW { i + 1}')
    for val in arr:
        print(f'{val:.2f}', end= '')
    print()
```

```python
tv.vocabulary_
```

```python
tv = TfidfVectorizer(max_features = 50, ngram_range = (2,2))
x = tv.fit_transform(corpus).toarray()
```
