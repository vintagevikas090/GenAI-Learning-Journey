### Lemmatization

###### Much Like Stemming . Resultant word is called 'Lemma'
###### Slower Than Stemming as each word is scanned against Wordnet Corpus Library words
###### POS is Req. [noun(n), ver(v), adjective(a), adverb(r)]
###### POS -> defines what each word should be treated as

```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lem = WordNetLemmatizer()
```

```python
lem.lemmatize('eating', pos = 'n') # treat eating as NOUN(default)
```

```python
lem.lemmatize('eating', pos = 'v') # treat eating as VERB
```

```python
lem.lemmatize('eating', pos = 'a') # treat eating as ADJECTIVE
```

```python
lem.lemmatize('eating', pos = 'r') # treat eating as ADVERB
```

```python
print('--- Lemmatization Output ---')
for word in words:
    print(word, "----> ", lem.lemmatize(word)) # as Noun
```

```python
print('--- Lemmatization Output ---')
for word in words:
    print(word, "----> ", lem.lemmatize(word, pos = 'v')) # as verb
```
