### POS Tag

###### POS tags stand for Parts of Speech tags. They are labels assigned to words in a sentence to identify the grammatical role each word plays.

###### For example, in the sentence:  “The cat is sleeping peacefully.”
###### “The” → Determiner | “cat” → Noun | “is” → Verb | “sleeping” → Verb | “peacefully” → Adverb

###### POS tagging is an important task in Natural Language Processing (NLP) because it helps computers understand sentence structure and meaning.

```python
### POS Tag -> NEEDS List of words as input
```

```python
sentence = 'This is Vikas Prajapat'
```

```python
import nltk
pos_tags = nltk.pos_tag(sentence.split())
pos_tags
```

### Named Entity Relationship

###### Try to find the entity (person, place, date, organisation etc.) from sentence or paragrapah
###### Taked tagged words (POS Tagged List of Words eg. pos_tags) AS INPUT

```python
# FROM paragraph to NER

# 1. para to words
# 2. words to pos tags
# 3. pos tags to NER
```

```python
corpus = """Artificial Intelligence is transforming the way humans interact with technology. 
From virtual assistants and recommendation systems to self-driving cars and medical diagnosis tools, 
AI is becoming an essential part of modern life. Generative AI, a branch of artificial intelligence, 
focuses on creating new content such as text, images, music, and code by learning patterns from large datasets. 
Large Language Models (LLMs) like ChatGPT can understand context, answer questions, summarize information, 
and even assist in software development. As AI continues to evolve, it is creating new opportunities in education, 
healthcare, business, and research while also raising important discussions about ethics, privacy, and responsible usage."""
```

```python
from nltk.tokenize import wordpunct_tokenize
words = wordpunct_tokenize(corpus)
print(words)
```

```python
tagged_elements = nltk.pos_tag(words)
print(tagged_elements)
```

```python
nltk.ne_chunk(tagged_elements).draw()
```
