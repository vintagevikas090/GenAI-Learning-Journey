### StopWords

###### Common words that usually carry little meaningful information for understanding the main context of text.
###### eg. [is, the, and, was, to, of etc.]

```python
from nltk.corpus import stopwords
```

```python
import nltk
nltk.download('stopwords')
```

```python
sw = stopwords.words('english')
```

```python
print(sw)
```

```python
corpus = """Artificial Intelligence is transforming the way humans interact with technology. 
From virtual assistants and recommendation systems to self-driving cars and medical diagnosis tools, 
AI is becoming an essential part of modern life. Generative AI, a branch of artificial intelligence, 
focuses on creating new content such as text, images, music, and code by learning patterns from large datasets. 
Large Language Models (LLMs) like ChatGPT can understand context, answer questions, summarize information, 
and even assist in software development. As AI continues to evolve, it is creating new opportunities in education, 
healthcare, business, and research while also raising important discussions about ethics, privacy, and responsible usage."""
```

```python
stopwords_in_corpus = []
for word in corpus.split():
    # print(word)
    if word in sw:
        stopwords_in_corpus.append(word)
```

```python
print(stopwords_in_corpus)
```

```python
import numpy as np
np.unique(stopwords_in_corpus)
```

```python
### Removing Stop words from our corpus -> keeping only meaningful words
```

```python
req_words = [word for word in corpus.split() if word not in sw]
corrected_corpus = ' '.join(req_words)
```

```python
corrected_corpus
```

```python
print('As' in corrected_corpus)
### The issue is that the stopwords are in lowercase and hence it can't remove this one
```

```python
req_words = [word for word in corpus.split() if word.lower() not in sw]
improved_corrected_corpus = ' '.join(req_words)
```

```python
improved_corrected_corpus
```

```python
print(len(corrected_corpus), '---', len(improved_corrected_corpus))
```

```python
print('As' in improved_corrected_corpus)
```
