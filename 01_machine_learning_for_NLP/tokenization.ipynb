### Tokenization Using NLTK

```python
corpus="""Hello, This is Vikas Prajapat's First notebook in GenAI course. It's going to be a long journey.
"""
```

```python
##  Tokenization
## paragraphs --> sentences
from nltk.tokenize import sent_tokenize
print(sent_tokenize(corpus))
```

```python
## Tokenization 
## Paragraph-->words
## sentence--->words
from nltk.tokenize import word_tokenize
print(word_tokenize(corpus))

# "  's  " is a single word. 
```

```python
from nltk.tokenize import wordpunct_tokenize
print(wordpunct_tokenize(corpus))

# "  's  " are two different words -> (" ' ", " s "). 
```

```python
from nltk.tokenize import TreebankWordTokenizer
tbwt = TreebankWordTokenizer()
print(tbwt.tokenize(corpus))

## doesn't Treats the '.' seperate word (except for the last word)
```
