##### Based on certain conditions, I may be required to call different tools. Hence cond routing is required 

```python
from typing import TypedDict
from langgraph.graph import StateGraph,START,END
```

```python
class TaskState(TypedDict):
    task: str
```

```python
def task_router(state):
    return {}

def search_agent(state):
    print("\nSEARCH AGENT")
    return {}

def code_agent(state):
    print("\nCODE AGENT")
    return {}

def database_agent(state):
    print("\nDATABASE AGENT")
    return {}

def general_agent(state):
    print("\nGENERAL AGENT")
    return {}
```

```python
### router function -> based on this, agent is called
def route_task(state):
    task = state["task"]

    if task == "search":
        return "search"

    if task == "code":
        return "code"

    if task == "database":
        return "database"

    return "general"
```

```python
builder = StateGraph(TaskState)

builder.add_node("router", task_router)

builder.add_node("search", search_agent)

builder.add_node("code", code_agent)

builder.add_node("database", database_agent)

builder.add_node("general", general_agent)
```

```python
builder.add_edge(START, "router")

builder.add_conditional_edges( "router", route_task) ### Adding the conditional edege -> will decide which tool to call

builder.add_edge("search", END)

builder.add_edge("code", END)

builder.add_edge("database", END)

builder.add_edge("general", END)
```

```python
graph = builder.compile()
```

```python
graph.invoke(
    {
        "task": "search"
    }
)
```

```python
graph.invoke(
    {
        "task": "code"
    }
)
```
