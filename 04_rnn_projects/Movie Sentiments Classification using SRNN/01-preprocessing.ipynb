```python
import pandas as pd
import re
import numpy as np
import tensorflow as tlf
from tensorflow.keras.layers import TextVectorization
```

# PREPROCESSING

```python
df = pd.read_csv('data/IMDB Dataset.csv')
```

```python
df.head()
```

```python
df['review'][0]
```

### REMOVING BR TAGS

```python
def remove_br_tag(text):
    return re.sub(r'<br\s*/*>', '', text)
```

```python
new_text = remove_br_tag(df['review'][0])
new_text
```

```python
df['review'] = df['review'].apply(remove_br_tag)
```

### REMOVING OTHER SPECIAL CHARs

```python
def remove_special_chars(text):
    # except for space, comma, fullstop
    return re.sub(r"[^a-zA-Z.,1-9'\s]", '', text)
```

```python
remove_special_chars(new_text)
```

```python
df['review'] = df['review'].apply(remove_special_chars)
```

### LOWERCASING THE TEXT

```python
df['review'] = df['review'].str.lower()
```

```python
df.head()
```

### No OF UNIQUE WORDS

```python
df['uniques'] = df['review'].apply(lambda text: len(set(text.split())))
```

```python
vocab_size = sum(df['uniques'])
vocab_size
```

```python
df.drop('uniques', axis = 1, inplace = True)
```

### Max VECTOR LENGTH

```python
### For padding, we need optimal vector len to pad extra zeros optimally
```

```python
df['len'] = df['review'].apply(lambda x: len(x.split()))
```

```python
max_len = np.percentile(df.len, 85)
int(max_len)
```

```python
df.drop('len', axis = 1, inplace = True)
```

```python
df['sentiment'] = df['sentiment'].str.lower()
df['sentiment'] = df['sentiment'].apply(lambda x: 1 if x == 'positive' else 0)
```
